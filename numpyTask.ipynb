```python
# increase salary by 10%
salary = np.array([50000,23000,12000])
inc_salary = salary + (salary * 0.10)
print(inc_salary)
```

```python
# decrease salary by 5%
dec_salary = salary - (salary * 0.10)
print(dec_salary)
```

```python
# square root
a = np.array([4,9,16])
print(np.sqrt(a))
```

```python
# find sine, cosine, tangent values
n = np.array([np.pi/2,np.pi/3,np.pi/6])
sine = np.sin(n)
print(sine)
cose = np.cos(n)
print(cose)
tane = np.tan(n)
print(tane)
```

```python
# find sec, cosec, cot,
print(1/sine)
print(1/cose)
print(1/tane)
```

```python
marks = np.array([67,50,65,34,40])
# add 5 grade marks
print(marks+5)

# average marks
print(np.average(marks))

#highest marks
print(np.max(marks))

# lowest marks
print(np.min(marks))

# marks >40
pass_std = np.sum(marks>40)
print(pass_std)
```

```python
# descibe function

def describe(arr):
    result = {
        "mean": np.mean(arr),
        "std": np.std(arr),
        "min": np.min(arr),
        "max": np.max(arr),
        "median": np.median(arr)
    }

    return result

marks = np.array([20,30,40,50])
summary =describe(marks)

for key, value in summary.items():
    print(f"{key}: {value}")
```
