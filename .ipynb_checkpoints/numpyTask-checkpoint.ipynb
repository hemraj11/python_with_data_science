```python
import numpy as np
```

```python
# array from 1 to 20
arr = np.arange(1,21)
print(arr)
```

```python
# array of 5 zeros
arr = np.zeros(5)
print(arr)
```

```python
# array of 5 ones
arr = np.ones(5)
print(arr)
```

```python
# 5 evenly spaced numbers between 0 and 1
arr= np.linspace(0,1,5)
print(arr)
```

```python
# multiply all the elements by3
nums = np.array([1,2,3,4])
print(nums)
mul=3*nums
print(mul)
```

```python
# add 10 to all numbers
add = nums + 10
print(add)
```

```python
# Find square of all the elements
print(nums**2)
```

```python
# square root
a = np.array([4,9,16])
print(np.sqrt(a))
```

```python
# find exponential values
print(np.exp(a))
```

```python
# find sine, cosine, tangent values
n = np.array([np.pi/2,np.pi/3,np.pi/6])
sine = np.sin(n)
print(sine)
cose = np.cos(n)
print(cose)
tane = np.tan(n)
print(tane)
```

```python
# find sec, cosec, cot,
print(1/sine)
print(1/cose)
print(1/tane)
```
