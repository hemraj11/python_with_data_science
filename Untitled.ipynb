```python
# universal functions ufuncs

arr=np.array([4,9,16])
result=np.sqrt(arr)
print(result)
print(np.exp(arr))
```

```python
# statistical operations
salary=np.array([2000,3000,1000,9000])
print(np.sum(salary))
print(np.mean(salary))
print(np.average(salary))
print(np.min(salary))
print(np.max(salary))
print(np.max(salary)-np.min(salary))
print(np.std(salary)) #SD
print(np.median(salary))
print(np.percentile(salary,20))
print(np.cumsum(salary))
print(np.cumprod(salary))
print(np.diff(salary))
```

```python
arr = np.array([
    [2,4,6],
    [3,7,8]
])

print(np.sum(arr,axis=1)) #row wise sum
print(np.sum(arr,axis=0)) #column wise sum
print(np.max(arr,axis=1))  #row wise max
print(np.max(arr,axis=0))  #column wise max
```

```python
marks = np.array([
    [20,20,40],  #student1
    [20,60,50],  #student2
    [23,34,45],  #student3
])

print(marks.mean(axis=1)) #avg marks of each students
print(marks.mean(axis=0)) #avg of marks of each subjects
```

```python
arr=np.array([10,20,30,40,50])
print(arr>20)
```

```python
marks = np.array([20,12,56,76,87])
print(marks[marks>32])
print(len(marks))
```

```python
result = np.where((marks>32) ,"Pass","Fail")
print(result)
result_sec = np.where((marks>32) & (marks<79),"Pass","Fail")
print(result_sec)
```
