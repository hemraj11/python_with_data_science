```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("salary.csv")
print(df)
```

```python
# select only salary col using loc
sal = df.loc[:, ["Salary"]]
print(sal)
```

```python
# select name and salary col using loc
n = df.loc[:,["Name", "Salary"]]
print(n)
```

```python
# select first 2 col using iloc
n = df.iloc[:,:2]
print(n)
```

```python
# select all rows and col using loc
n = df.loc[:,:]
print(n)
```

```python
# select dept and age using iloc
n = df.iloc[:, [1,3]]
print(n)
```

```python
# select 1st row and all col using loc
n = df.loc[0,:]
print(n)
```

```python
# select rows: 1 to 3 and col: Name,Salary using loc
n = df.loc[1:3,['Name','Salary']]
print(n)
```

```python
# select first 3 rows and first 2 col using iloc
n = df.iloc[:3,:2]
print(n)
```

```python
# select salary>45000, show name and salary
n= df.loc[df['Salary']>45000, ['Name','Salary']]
print(n)
```

```python
# select last col using iloc
n = df.iloc[:, -1]
print(n)
```

```python
# select col in reverse order
n= df.iloc[:, ::-1]
print(n)
```

```python
# select all col except age in pandas
n = df.iloc[:, :-1]
print(n)
```

```python
n = df.loc[:, df.columns != 'Age']
print(n)
```
