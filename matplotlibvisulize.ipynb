```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
x=[2,3,4,5]
y=[12,10,14,9]
plt.plot(x,y)
plt.show()
```

```python
days=["Sun","Mon","Tue"]
sales=[2000,1000,4000]
plt.plot(days,sales, marker='o',linestyle='--',color='blue', markerfacecolor='red', markersize=10)
plt.title("Sales over the week")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.show()
```

```python
subject=['Math','Science','English']
marks=[67,78,34]
plt.bar(subject,marks,color='red')
plt.title("Students Bar Graph", color='teal',size=20)
plt.xlabel("Subject", color='red',size=15)
plt.ylabel("Marks", color='red',size=15)
plt.show()
```

```python
x=[1,2,3,4]
y1=[10,4,45,6]
y2=[15,42,35,4]
plt.plot(x,y1)
plt.plot(x,y2)
plt.grid(axis='x',color='teal', linestyle='--')
plt.show()
```

```python
df=pd.read_csv("sales_data.csv")
print(df)
```

```python
ram_data=df[df['Name']=="Ram"]
plt.plot(ram_data['Month'],ram_data['Sales'])
plt.title("Ram's Slaes Data")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
df['Sales']= df['Sales'].replace('400s','400')
df['Sales']= pd.to_numeric(df['Sales'])
dept_sales = df.groupby("Dept")["Sales"].sum()
plt.bar(dept_sales.index,dept_sales.values)
plt.title("Total Sales by Departments")
plt.xlabel("Departments")
plt.ylabel("Sales")
plt.show()
```
